```python
from langchain_ollama import ChatOllama

print("LangChain + Ollama OK")
```

```python
llm = ChatOllama(model="qwen3:8b", temperature=0)

print("Model configured")
```

```python
response = llm.invoke("Explain what an API is in one simple sentence.")

print(response.content)
```

```python
import time

start = time.perf_counter()

response = llm.invoke("Reply with exactly: OK")

elapsed = time.perf_counter() - start

print(response.content)
print(f"Time: {elapsed:.2f} seconds")
print(response.response_metadata)
```

```python
from langchain_learning.chains import get_basic_chain

chain = get_basic_chain()

for chunk in chain.stream({"question": "Explain TCP/IP in three short sentences."}):
    print(chunk, end=" ", flush=True)
```
